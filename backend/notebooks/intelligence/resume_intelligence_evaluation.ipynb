```python
from app.services.resume_service import extract_resume_document
from app.ai.analyzers.resume_analyzer import analyze_resume
from app.ai.analyzers.intelligence.ats_analyzer import (
    analyze_resume_ats,
)

document = extract_resume_document(
    str(PDF_PATH)
)

resume_analysis = analyze_resume(
    document.text
)

ats_result = analyze_resume_ats(
    resume_document=document,
    resume_analysis=resume_analysis,
)

print("ATS Score:", ats_result["overall_score"])
print("Label:", ats_result["score_label"])
```

```python
for name in [
    "parsing_reliability",
    "section_structure",
    "terminology_quality",
    "content_consistency",
    "ats_risk_score",
]:
    dimension = ats_result[name]

    print(
        f"\n{name}"
        f"\nScore: {dimension['score']}/100"
        f"\nWeight: {dimension['weight']}%"
        f"\nRationale: {dimension['rationale']}"
    )
```

```python
print("Overall:", ats_result["overall_score"])

for name in [
    "parsing_reliability",
    "section_structure",
    "terminology_quality",
    "content_consistency",
    "ats_risk_score",
]:
    item = ats_result[name]
    print(
        name,
        item["score"],
        item["weight"],
    )
```

```python
weighted_total = sum(
    ats_result[name]["score"]
    * ats_result[name]["weight"]
    for name in [
        "parsing_reliability",
        "section_structure",
        "terminology_quality",
        "content_consistency",
        "ats_risk_score",
    ]
) / 100

print("Expected:", weighted_total)
```

```python
print("ISSUES")
print("=" * 60)

for issue in ats_result["issues"]:
    print(f"\nCategory: {issue['category']}")
    print(f"Title: {issue['title']}")
    print(f"Severity: {issue['severity']}")
    print(f"Source: {issue['source']}")
    print(f"Description: {issue['description']}")
```

```python
print("TERMINOLOGY")
print("=" * 60)

print(
    "Score:",
    ats_result["terminology"]["score"],
)

print("\nNormalized terms:")
for term in ats_result["terminology"]["normalized_terms"]:
    print("-", term)

print("\nInconsistencies:")
for item in ats_result["terminology"]["inconsistencies"]:
    print("-", item)
```

```python
print("RECOMMENDATIONS")

for item in ats_result["recommendations"]:
    print(
        f"\nPriority: {item['priority']}"
        f"\nCategory: {item['category']}"
        f"\nTitle: {item['title']}"
        f"\nDescription: {item['description']}"
    )
```
